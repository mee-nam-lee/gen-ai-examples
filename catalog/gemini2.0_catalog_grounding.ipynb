### Import libraries


```python
from IPython.display import HTML, Markdown, display
from google import genai
from google.genai.types import (
    GenerateContentConfig,
    GoogleSearch,
    Part,
    Tool,
)
```

```python
import os
from dotenv import load_dotenv

load_dotenv()
PROJECT_ID = os.environ.get("PROJECT_ID")
LOCATION = os.environ.get("REGION", "us-central1")
```

#### Use a Google Cloud Project


```python
client = genai.Client(vertexai=True, project=PROJECT_ID, location=LOCATION)
```

### Load the Gemini 2.0 Flash model

Learn more about all [Gemini models on Vertex AI](https://cloud.google.com/vertex-ai/generative-ai/docs/learn/models#gemini-models).

```python
MODEL_ID = "gemini-2.0-flash-001"  # @param {type: "string"}
```

```python
chat = client.chats.create(model=MODEL_ID)
```

```python
system_instruction = """
    You are a product catalog manager for ecommerce. Answer the given questions in Korean..
"""
```

```python
config = GenerateContentConfig(
    temperature = 0.9,
    top_p = 0.95,
    max_output_tokens = 8192,
    response_modalities = ["TEXT"],
    system_instruction=system_instruction   
)
```

<img src=./coat1.png width="10%">

```python
with open("coat1.png", "rb") as f:
    image = f.read()

product="KARRA 체크라이닝럭셔리코트_A4F0160"
product_description="간절기 시즌에 필수로 소장하기 좋은 트렌치 코드입니다.무릎까지 내려오는 롱한 기장감으로 안정감 있는 착용을 도와주면서 시크해 보이는 무드를 연출시켜 줍니다. 안쪽 체크무늬 패턴의 안감으로 고급스러움을 더해 주었습니다. 더블 버튼 잠금으로 깔끔한 핏을 연출시켜줍니다."
prompt = f"Categorize {product} into 4 depth categories based on the given product image and product descriptions, {product_description} ."
response = chat.send_message([
        Part.from_bytes(data=image, mime_type="image/png"),
        prompt,
    ],
    config=config)

display(Markdown(response.text))
```

# Product Description 없이 질문 (Grounding with Google Search)

```python
google_search_tool = Tool(google_search=GoogleSearch())
grounding_config = GenerateContentConfig(
    temperature = 0.9,
    top_p = 0.95,
    max_output_tokens = 8192,
    response_modalities = ["TEXT"],
    system_instruction=system_instruction,
    tools=[google_search_tool]   
)
```

```python
product="나이키 운동화 에어 모나크 4 (415445-102)"
prompt = f"Categorize {product} into 4 depth categories based on the given product image like : 1st category / 2nd category / 3rd category / 4th category."

response = client.models.generate_content(
    model=MODEL_ID,
    contents=[Part.from_bytes(data=image, mime_type="image/png"),
        prompt,],
    config=grounding_config
)

display(Markdown(response.text))

#HTML(response.candidates[0].grounding_metadata.search_entry_point.rendered_content)

print("Grounding : ")
for chunk in response.candidates[0].grounding_metadata.grounding_chunks:
    display(HTML(f"<a href='{chunk.web.uri}'>{chunk.web.title}</a>"))
```

```python
prompt = f"Please recreate the description by referring to the product image and extract keywords that best describe the features of this product."

response = client.models.generate_content(
    model=MODEL_ID,
    contents=[Part.from_bytes(data=image, mime_type="image/png"),
        prompt,],
    config=grounding_config
)

display(Markdown(response.text))
print("Grounding : ")

HTML(response.candidates[0].grounding_metadata.search_entry_point.rendered_content)

for chunk in response.candidates[0].grounding_metadata.grounding_chunks:
    display(HTML(f"<br/><br/><a href='{chunk.web.uri}'>{chunk.web.title}</a>"))
```
