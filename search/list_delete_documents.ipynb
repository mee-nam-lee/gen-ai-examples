```python
PROJECT_ID = os.environ.get('PROJECT_ID')
LOCATION=os.environ.get('LOCATION')
DATASTORE_ID=os.environ.get('DATASTORE_ID')
GS_BUCKET=os.environ.get('GS_BUCKET')
```

```python
from google.cloud import discoveryengine_v1 as discoveryengine
```

```python
from pprint import pprint 

def list_documents(project_id: str, location: str, data_store_id: str) -> List[Dict[str, str]] | None:
       # Create a client
    client = discoveryengine.DocumentServiceClient()

    # Initialize request argument(s)
    request = discoveryengine.ListDocumentsRequest(
        parent=f"projects/{project_id}/locations/{location}/collections/default_collection/dataStores/{data_store_id}/branches/0",
    )

    document_list = []
    
    # Make the request
    page_result = client.list_documents(request=request)

    # Handle the response
    for response in page_result:
        document_info = {}
        document_info['id'] = response.id
        document_info['name'] = response.name
        document_info['uri'] = response.content.uri
        document_list.append(document_info)
   
    pprint(document_list)
    return document_list
        
```

```python
list_documents(PROJECT_ID, LOCATION, DATASTORE_ID)
```

```python
def get_document_by_name(project_id: str, location: str, data_store_id: str, document_name: str) -> Dict[str, str] | None:
 
    client = discoveryengine.DocumentServiceClient()

    # Initialize request argument(s)
    request = discoveryengine.GetDocumentRequest(
        name=document_name
    )

    # Make the request
    try:
        response = client.get_document(request=request)
        print(response)
    except Exception as e:
        print(f"Exception : {e}")
```

```python
def get_document_by_uri(project_id: str, location: str, data_store_id: str, document_gcs_uri: str) -> Dict[str, str] | None:
 
    client = discoveryengine.DocumentServiceClient()

    # Initialize request argument(s)
    request = discoveryengine.ListDocumentsRequest(
        parent=f"projects/{project_id}/locations/{location}/collections/default_collection/dataStores/{data_store_id}/branches/0",
    )

    document_list = []
    
    # Make the request
    page_result = client.list_documents(request=request)

    # Handle the response
    for response in page_result:
        document_info = {}
        document_info['id'] = response.id
        document_info['name'] = response.name
        document_info['uri'] = response.content.uri
        document_list.append(document_info)
   
    #pprint(document_list)

    for doc in document_list:
        if doc['uri'] == document_gcs_uri:
            return doc['name'] 
```
